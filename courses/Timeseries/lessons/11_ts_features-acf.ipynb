## ACF features

All the autocorrelations of a series can be considered features of that series. We can also summarise the autocorrelations to produce new features. The `feat_acf()` computes a selection of seven autocorrelation features:

- $\rho_1$: the first autocorrelation coefficient from the original data
- $\sum_{k=1}^{10} \rho_k^2$: the sum of squares of the first ten autocorrelation coefficients from the original data
- $\rho_1^{(1)}$: the first autocorrelation coefficient from the differenced data ($\Delta x_t$)
- $\sum_{k=1}^{10} \left(\rho_k^{(1)}\right)^2$: the sum of squares of the first ten autocorrelation coefficients from the differenced data
- $\rho_1^{(2)}$: the first autocorrelation coefficient from the twice differenced data ($\Delta^2 x_t$)
- $\sum_{k=1}^{10} \left(\rho_k^{(2)}\right)^2$: the sum of squares of the first ten autocorrelation coefficients from the twice differenced data
- For seasonal data, $\rho_m$: the autocorrelation coefficient at the first seasonal lag ($m$), where $m$ is the seasonal period

**Differencing** is what makes this bundle powerful. Instead of the raw level $y_t$, we look at changes:

- **first difference** $y_t - y_{t-1}$ — the quarter-to-quarter change;
- **second difference** — the change *of* the change;
- **seasonal difference** $y_t - y_{t-4}$ — the year-over-year change for the same quarter.

The autocorrelations of this double differenced series may provide useful information.

We will not use ACF features direclty, rather, we will go straight to higher level features: STL features (Next).
